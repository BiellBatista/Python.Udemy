```python
#coding: utf-8

from kivy.app import App
from kivy.uix.button import Button
from kivy.uix.label import Label
from kivy.interactive import InteractiveLauncher
from kivy.lang import Builder
from kivy.config import Config
#força a aplicação a não inciar em tela cheia
Config.set("graphics", "fullscreen", "0")

janela = None
glayout = None
```

```python
class JanelaApp(App):
    pass

janela = JanelaApp()
ji = InteractiveLauncher(janela)
ji.run()
```

```python
kvcode = """
StackLayout:
    orientation: "tb-lr"
    padding: 50    
"""
# limpando o gerenciado de layout para que seja interativo
if(janela.root):
    janela.root_window.remove_widget(janela.root)
    janela.root = None
    glayout = None

janela.root = glayout = Builder.load_string(kvcode)
janela.root_window.add_widget(glayout)
```

```python
def add_lb(**args):
    lb = Label(text="Estudando a classe label",
                  size_hint_y = None,
                   height = 50, #altura de cada label
                   **args
              )
    glayout.add_widget(lb)
    return lb
```

```python
#font_size determina o tamanho da fonte
add_lb().font_zie = 10
```

```python
#font_name determina a fonte do texto
x = add_lb()
x.font_name = "consola"
```

```python
#bold determina se a fonte dever negrito
add_lb().bold = True
```

```python
add_lb().italic = True
```

```python
add_lb(font_size=20).color = .1, .2, .3, 1
```

```python
add_lb(font_size=30).disabled = True
```

```python
txt = """aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd
aaa bb cc ddd"""
```

```python
x = add_lb()
x.text = txt
```

```python
lb = add_lb()
lb.text = txt
lb.line_height = 3
```

```python
lb = add_lb()
lb.text = txt
lb.max_line = 3 #o label pode exibir no máximo 3 linhas, de baixo pra cima
```
