```python
from kivy.app import App
from kivy.uix.button import Button
from kivy.uix.label import Label
from kivy.interactive import InteractiveLauncher
from kivy.lang import Builder
from kivy.config import Config
#força a aplicação a não inciar em tela cheia
Config.set("graphics", "fullscreen", "0")

janela = None
glayout = None
```

```python
class JanelaApp(App):
    pass

janela = JanelaApp()
ji = InteractiveLauncher(janela)
ji.run()
```

```python
kvcode = """
StackLayout:
    orientation: "tb-lr"
    padding: 50    
"""
# limpando o gerenciado de layout para que seja interativo
if(janela.root):
    janela.root_window.remove_widget(janela.root)
    janela.root = None
    glayout = None

janela.root = glayout = Builder.load_string(kvcode)
janela.root_window.add_widget(glayout)
```

```python
def add_bt(**args):
    bt = Button(text="Estudando a classe Button.",
                size_hint_y=None,
                height = 50, #altura de cada botão
                markup=True,
                **args)
    glayout.add_widget(bt)
    return bt
```

```python
add_bt().font_size = 30
```

```python
add_bt().font_name = "consola"
```

```python
add_bt(font_size=22).bold = True
```

```python
add_bt(font_size=22).italic = True
```

```python
add_bt().color = .9, .5, .3, 1
```

```python
bt = add_bt()
bt.markup = True
bt.text = "Esse [b]texto[/b] possui formatação"
```

```python
#on_press
def click(bt):
    bt.text = "on_press"
def fim_click(bt):
    bt.text = "on_realese"
    
bt = add_bt()
bt.bind(on_press=click) #o método bind() liga um evento da classe a uma função fora dela
bt.bind(on_release=fim_click)
```
