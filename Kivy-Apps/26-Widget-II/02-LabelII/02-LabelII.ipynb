```python
#coding: utf-8

from kivy.app import App
from kivy.uix.button import Button
from kivy.uix.label import Label
from kivy.interactive import InteractiveLauncher
from kivy.lang import Builder
from kivy.config import Config
#força a aplicação a não inciar em tela cheia
Config.set("graphics", "fullscreen", "0")

janela = None
glayout = None
```

```python
class JanelaApp(App):
    pass

janela = JanelaApp()
ji = InteractiveLauncher(janela)
ji.run()
```

```python
kvcode = """
StackLayout:
    orientation: "tb-lr"
    padding: 50    
"""
# limpando o gerenciado de layout para que seja interativo
if(janela.root):
    janela.root_window.remove_widget(janela.root)
    janela.root = None
    glayout = None

janela.root = glayout = Builder.load_string(kvcode)
janela.root_window.add_widget(glayout)
```

```python
def add_lb(**args):
    lb = Label(size_hint_y = None,
                   font_size=20,
                   height = 50, #altura de cada label
                   **args
              )
    glayout.add_widget(lb)
    return lb
```

```python
#o argumento markup informa que uma determinada cadeia de string possui marcadores
add_lb(markup=True).text = "Estudando a [b]classe[/b] Label"
```

```python
[b][/b] #definindo uma tag para palavra negrito
[i][/i] #definindo uma tag para palavra itálico
[u][/u]
[size=TAMANHO][/size] #muda o tamanho de um treco
[color=CORemHEXADECIMAL][/size] #muda a cor de um trecho
[sup][/sup] #faz com que determinado trecho de código seja exibido acima dos demais
[sub][/sub] #faz com que determinado trecho de código seja exibido abaixo dos demais
```
