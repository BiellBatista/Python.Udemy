```python
#coding: utf-8

from kivy.app import App
from kivy.uix.button import Button
from kivy.interactive import InteractiveLauncher
from kivy.lang import Builder
from kivy.config import Config
#força a aplicação a não inciar em tela cheia
Config.set("graphics", "fullscreen", "0")

janela = None
glayout = None
```

```python
class JanelaApp(App):
    pass

janela = JanelaApp()
ji = InteractiveLauncher(janela)
ji.run()
```

```python
kvcode = """
GridLayout:
    cols: None #colunas terá a mesma quantidade da linha
    rows: 3 #linhas
    
    row_default_height: 50 #cada linha terá 50 pixels de altura
    row_force_default: True #faz a linha de cima ser verdadeira
    
    col_default_width: 50 #cada coluna terá 50 pixel de largura
    col_force_default: True
    
    Button:
        text: "A"
        size_hint: .3, None
    Button:
        text: "B"
    Button:
        text: "C"
    Button:
        text: "D"
        size_hint: .3, None
    Button:
        text: "E"
    Button:
        text: "F"
    Button:
        text: "G"
        size_hint: .3, None
"""
# limpando o gerenciado de layout para que seja interativo
if(janela.root):
    janela.root_window.remove_widget(janela.root)
    janela.root = None
    glayout = None

janela.root = glayout = Builder.load_string(kvcode)
janela.root_window.add_widget(glayout)
```

```python
glayout.add_widget(Button(text="x"))
```
